```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import load_wine
wine = load_wine()
wine_df = pd.DataFrame(wine.data, columns=wine.feature_names)
wine_df.head()
```

```python
wine_df['target'] = wine.target
```

```python
wine_df.groupby('target').mean().T
```

```python
wine_df.groupby('target').mean().T.plot(kind='bar')
```

```python
wine_df.columns
```
